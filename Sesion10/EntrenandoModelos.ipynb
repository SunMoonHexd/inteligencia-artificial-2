El aprendizaje automático supervisado utiliza la optimización matemática para encontrar los mejores parámetros de un modelo a partir de datos históricos.

---

### 1. Regresión Lineal y Polinomial

* **Regresión Lineal:** Modela la relación entre variables predictoras ($X$) y una variable continua ($y$) mediante una combinación lineal de los parámetros ($\theta$).

$$\hat{y} = \theta_0 + \theta_1 x_1 + \theta_2 x_2 + \dots + \theta_n x_n$$


* **Regresión Polinomial:** Extiende la regresión lineal para capturar relaciones no lineales agregando potencias de las características originales (p. ej., $x^2, x^3$). Sigue siendo un modelo lineal respecto a sus coeficientes $\theta$.

---

### 2. Entrenamiento con el Método del Gradiente (Gradient Descent)

Para entrenar un modelo, se define una **función de costo** (como el Error Cuadrático Medio, MSE) que mide la diferencia entre las predicciones y los valores reales:

$$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2$$

El **Descenso del Gradiente** ajusta iterativamente los parámetros $\theta$ moviéndose en la dirección opuesta al gradiente de la función de costo:

$$\theta_j := \theta_j - \alpha \frac{\partial}{\partial \theta_j} J(\theta)$$

Donde $\alpha$ es la **tasa de aprendizaje** (*learning rate*), la cual controla el tamaño del paso en cada iteración.

---

### 3. Curvas de Aprendizaje (Overfitting vs. Underfitting)

Las curvas de aprendizaje grafican el rendimiento del modelo (error o *accuracy*) en los conjuntos de entrenamiento y validación a medida que aumenta el tamaño de los datos o las iteraciones.

* **Bajo ajuste (Underfitting / Sesgo alto):** Los errores de entrenamiento y validación son altos y cercanos entre sí. El modelo es demasiado simple.
* **Sobreajuste (Overfitting / Varianza alta):** El error de entrenamiento es bajo, pero el error de validación es muy alto. El modelo memorizó el ruido.
* **Buen ajuste:** Ambos errores convergen a un valor bajo con una brecha pequeña entre ellos.

---

### 4. Regularización

Técnica utilizada para prevenir el sobreajuste penalizando la complejidad de los coeficientes del modelo en la función de costo:

* **Lasso (L1):** Añade la suma del valor absoluto de los coeficientes:

$$J_{L1}(\theta) = J(\theta) + \lambda \sum_{j=1}^{n} \vert{}\theta_j\vert{}$$



*Efecto:* Puede reducir coeficientes exactamente a cero, realizando selección de características.
* **Ridge (L2):** Añade la suma del cuadrado de los coeficientes:

$$J_{L2}(\theta) = J(\theta) + \lambda \sum_{j=1}^{n} \theta_j^2$$



*Efecto:* Encoje los coeficientes hacia cero sin eliminar ninguno por completo.

*(Donde $\lambda$ es el hiperparámetro de fuerza de regularización).*

---

### 5. Clasificación

A diferencia de la regresión, la clasificación predice etiquetas discretas (por ejemplo, spam / no spam o clases múltiples).

* **Regresión Logística:** Aplica la función sigmoide $\sigma(z) = \frac{1}{1 + e^{-z}}$ sobre una combinación lineal para transformar salidas continuas en probabilidades entre $0$ y $1$.
* **Función de costo:** Utiliza la pérdida logarítmica (*Log Loss* o *Cross-Entropy Loss*) en lugar del MSE para garantizar que la superficie de costo sea convexa y optimizable mediante gradiente descendente.
