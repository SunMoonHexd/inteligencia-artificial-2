## 5. Funciones y Grafos

## 1. Inspiración Biológica

* **Corteza Visual Humana:** Las redes neuronales convolucionales (CNN) se inspiran en cómo los ojos y el cerebro procesan la luz a través de campos receptivos simples y complejos.


* **Jerarquía de Aprendizaje:** Al igual que el cerebro humano, la red empieza detectando trazos primarios (líneas, bordes, contrastes) y los combina progresivamente hasta reconocer patrones y conceptos complejos.



## 2. Capas y Filtros

* **Convolución:** Operación fundamental donde matrices pequeñas (filtros/kernels) se desplazan sobre la imagen para extraer características visuales clave.


* **Filtros de Características:**
* *Primeras capas:* Filtran bordes, esquinas y colores.


* *Capas intermedias:* Reconocen texturas, formas geométricas y partes de objetos (ojos, ruedas, ventanas).


* *Capas profundas:* Identifican objetos completos o rostros.





## 3. Agrupación (Pooling)

* **Reducción de Dimensión:** Operaciones como *Max Pooling* o *Average Pooling* comprimen la información relevante conservando los datos clave y descartando detalles superfluos.


* **Invariancia a la Traslación:** Permite que la red reconozca un objeto sin importar en qué posición o ángulo de la imagen se encuentre presente.



## 4. Arquitecturas Famosas

* **Evolución Histórica:** Redes emblemáticas que marcaron hitos en el rendimiento de percepción visual:


* *LeNet-5:* Introductora de la convolución para lectura de caracteres.


* *AlexNet:* Revolución del aprendizaje profundo (*Deep Learning*) con GPU.


* *VGG / Inception:* Redes profundas que optimizan el uso de filtros a distintas escalas.


* *ResNet:* Introducción de conexiones residuales (*skip connections*) para entrenar cientos de capas sin degradar el gradiente.





## 5. Modelos Preentrenados

* **Transfer Learning (Aprendizaje por Transferencia):** Reutilización del conocimiento de modelos previamente entrenados en grandes bases de datos (como ImageNet).


* **Ajuste Fino (*Fine-Tuning*):** Reaprovechamiento de los filtros generales para adaptar la red a un problema o dominio específico con pocos datos de entrada.



## 6. Detección de Objetos

* **Localización y Clasificación:** Supera la clasificación simple al identificar *qué* hay en la imagen y *dónde* está mediante cajas delimitadoras (*bounding boxes*).


* **Arquitecturas Modernas:** Modelos de una sola etapa (como **YOLO** o **SSD**) y de dos etapas (como **Faster R-CNN**) para análisis y percepción en tiempo real.
