## 1. Orden del caos

En datos y Machine Learning, "ordenar el caos" se refiere a **encontrar patrones, estructuras o agrupaciones implícitas en grandes volúmenes de datos no estructurados o ruidosos** sin intervención humana directa. Es la premisa fundamental sobre la que se construyen los algoritmos de análisis exploratorio.

---

## 2. Aprendizaje no supervisado y datos no supervisados

* **Datos no supervisados:** Son conjuntos de datos que carecen de etiquetas o respuestas conocidas (no hay una variable objetivo $y$).
* **Aprendizaje no supervisado:** Algoritmos que analizan estos datos etiquetados para descubrir por sí mismos la estructura subyacente.
* **Clustering (Agrupamiento):** Divide los datos en grupos con características similares (ej. K-Means, DBSCAN).
* **Reducción de dimensionalidad:** Simplifica el número de variables manteniendo la mayor información posible (ej. PCA, t-SNE).



---

## 3. Aprendizaje activo (Active Learning)

Es una rama del aprendizaje semi-supervisado donde **el algoritmo interactúa activamente con un usuario/experto humano para etiquetar solo los datos más informativos**.

* **Objetivo:** Minimizar el costo y tiempo de etiquetado manual.
* **Mecanismo:** El modelo entrena con pocos datos etiquetados, identifica los puntos no etiquetados sobre los que tiene mayor incertidumbre y le pide al humano que solo etiquete esos casos críticos.

---

## 4. Detección de anomalías y novedades

Ambas buscan identificar datos que se desvían drásticamente del comportamiento "normal", pero con un enfoque distinto:

| Concepto | Definición | Contexto |
| --- | --- | --- |
| **Detección de anomalías** (*Outlier Detection*) | Encuentra datos atípicos o "contaminados" que **ya existen dentro del conjunto de entrenamiento**. | Detectar fraude en transacciones históricas. |
| **Detección de novedades** (*Novelty Detection*) | El modelo se entrena **solo con datos limpios/normales** y detecta si un *nuevo* dato entrante pertenece o no a esa distribución conocida. | Detectar fallas inéditas en motores recién salidos de fábrica. |

---

## 5. Mezclas Gaussianas (GMM - Gaussian Mixture Models)

Es un método probabilístico de clustering que asume que todos los puntos de datos son generados por una combinación (mezcla) de un número finito de **distribuciones Gaussianas (normales)** con parámetros desconocidos.

* **Soft Clustering:** A diferencia de K-Means (donde un punto pertenece o no a un cluster), GMM asigna a cada punto una **probabilidad** de pertenecer a cada distribución.
* **Flexibilidad:** Permite formar clusters elípticos de diferentes tamaños y orientaciones.
* Se optimiza comúnmente mediante el algoritmo **EM (Expectation-Maximization)**.

---

## 6. Criterios de selección de modelos (BIC y AIC)

Al usar modelos probabilísticos como las Mezclas Gaussianas, surge la pregunta: *¿Cuántos clusters o componentes $k$ debemos elegir?* Para responder sin caer en sobreajuste (*overfitting*), se utilizan criterios de información penalizada:

* **AIC (Akaike Information Criterion):** Evalúa qué tan bien se ajusta el modelo a los datos, penalizando el número de parámetros. Tiende a preferir modelos un poco más complejos.
* **BIC (Bayesian Information Criterion):** Similar a AIC, pero aplica una **penalización más severa** por la complejidad del modelo (depende del tamaño de la muestra).

> **Regla general:** Al comparar varios valores de $k$, se selecciona el modelo que minimice el valor de **AIC** o **BIC**.
