Las redes neuronales profundas sufren de problemas como la inestabilidad del gradiente al agregar más capas. Para entrenarlas de manera estable y eficiente, se emplean diversas técnicas sobre la inicialización de pesos, funciones de activación, normalización y reuso de conocimiento.

---

## 1. El Reto del Gradiente Inestable

Durante la retropropagación (backpropagation), la regla de la cadena multiplica las derivadas parciales de cada capa para calcular el gradiente de la función de pérdida con respecto a los pesos iniciales.

* **Gradiente Desvaneciente (*Vanishing Gradient*):** Ocurre cuando los gradientes se vuelven extremadamente pequeños al multiplicarse capa tras capa. Las capas profundas dejan de actualizar sus pesos y el entrenamiento se frena por completo.
* **Gradiente Explosivo (*Exploding Gradient*):** Los gradientes crecen exponencialmente en cada capa, volviéndose extremadamente grandes ($NaN$). Esto produce actualizaciones drásticas en los pesos que desestabilizan el modelo.

---

## 2. Inicialización de Pesos

Si todos los pesos se inicializan en cero, las neuronas aprenden exactamente las mismas características (ruptura de simetría fallida). Para evitar la inestabilidad del gradiente, la varianza de las salidas de una capa debe ser igual a la varianza de sus entradas.

* **Inicialización Xavier / Glorot:** Diseñada para funciones de activación simétricas centradadas en cero (como $\text{tanh}$ o la sigmoide). Mantiene constante la varianza de la señal a través de la red.
* **Inicialización He / Kaiming:** Optimizada para la función de activación $\text{ReLU}$ y sus variantes, ajustando la escala de la varianza para considerar que $\text{ReLU}$ anula la mitad de las entradas.

---

## 3. Funciones de Activación Modernas

Las funciones tradicionales como la **Sigmoide** o la **Tangente Hiperbólica ($\text{tanh}$)** se saturaban fácilmente en sus extremos, haciendo que sus derivadas cayeran casi a cero y provocando el desvanecimiento del gradiente.

| Función | Definición | Ventajas | Desventajas |
| --- | --- | --- | --- |
| **ReLU** *(Rectified Linear Unit)* | $f(x) = \max(0, x)$ | Cómputo rápido y no se satura para valores positivos. | **Dying ReLU:** Las neuronas cuyos valores caen bajo cero mueren permanentemente al tener gradiente 0. |
| **Leaky ReLU** | $f(x) = \max(\alpha x, x)$ con $\alpha \approx 0.01$ | Asigna una pendiente muy pequeña a las entradas negativas para evitar que la neurona muera. | El parámetro $\alpha$ debe definirse previamente de forma manual. |
| **ELU** *(Exponential Linear Unit)* | $f(x) = x \text{ si } x > 0 \text{ else } \alpha(e^x - 1)$ | Suave en todas sus partes; acelera la convergencia al acercar la media de las activaciones a cero. | Requiere el cómputo de la función exponencial, lo cual es más costoso numéricamente. |
| **GELU / Swish** | $x \cdot \Phi(x)$ / $x \cdot \sigma(\beta x)$ | Funciones suaves y no monótonas utilizadas ampliamente en la arquitectura Transformer y LLMs. | Mayor costo computacional. |

---

## 4. Normalización

A medida que los pesos cambian durante el entrenamiento, la distribución de las entradas de cada capa intermedia cambia continuamente, un fenómeno conocido como *Internal Covariate Shift*. Las técnicas de normalización reescalan las activaciones internas.

* **Batch Normalization (BN):** Normaliza las activaciones de un minilote (*mini-batch*) para que tengan media 0 y varianza 1, aplicando luego una escala y desplazamiento aprendibles ($\gamma$ y $\beta$). Permite el uso de tasas de aprendizaje más altas y actúa como regularizador.
* **Layer Normalization (LN):** Normaliza a través de las características de un solo ejemplo individual en lugar del lote. Se utiliza ampliamente en procesamiento de lenguaje natural (NLP) y arquitecturas Transformer.

---

## 5. Transferencia de Aprendizaje y Preentrenamiento

Aprovechar el conocimiento extraído de un modelo ya entrenado reduce drásticamente el tiempo de cómputo y la necesidad de grandes volúmenes de datos.

* **Preentrenamiento no supervisado (*Greedy Layer-wise*):** Método histórico utilizado antes del desarrollo de BN e inicializaciones modernas. Entrenaba la red capa por capa utilizando codificadores automáticos (*Autoencoders*) o Máquinas de Boltzmann Restringidas (RBMs) de forma no supervisada antes del ajuste fino (*fine-tuning*).
* **Transfer Learning (Transferencia de Aprendizaje):** Toma una red neuronal profunda preentrenada en un conjunto de datos masivo (como ImageNet) y la reutiliza para una tarea específica.
* **Extracción de características:** Se congelan los pesos de las capas inferiores/intermedias y solo se entrena una nueva capa clasificadora al final.
* **Fine-Tuning:** Se ajustan los pesos de toda la red (o de las últimas capas) con una tasa de aprendizaje muy baja para adaptar las representaciones aprendidas al dominio objetivo.
